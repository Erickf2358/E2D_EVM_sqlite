```python
import pandas as pd
import sqlite3
from datetime import timedelta  # <-- needed for date_range with Timedelta or you can use freq='D'

conn = sqlite3.connect(r'C:\Users\efloresmendoza\OneDrive - John Holland Group\2024\140_Erick\01_database\01.-COST ACCOUNTING\ROsample.db')

# Define the SQL query
query_03 = """
SELECT *
FROM Project_RO_table

"""

df = pd.read_sql_query(query_03, conn)
conn.close()
df

df_project = df[[
 'Activity',
 'Activity Name',
 'Planned Start',
 'Planned Finish',
 'Cost',
 'Cost account']].copy()
df.head()
df
df_act = df_project.rename(columns={
    'Activity': 'Activity_ID',
    'Cost account':'COA',
    'Activity Name': 'Activity_Description',
    'Planned Start': 'Start_Date',
    'Planned Finish': 'End_Date',
    'Cost': 'BAC',
})  

# Set the start of the upcoming month
df_act ['Today'] = '01/05/2026'

public_holidays = pd.to_datetime([
    '2026-01-01',  # New Year’s Day
    '2026-04-02',  # Maundy Thursday
    '2026-04-03',  # Good Friday
    '2026-05-01',  # Labour Day
    '2026-06-07',  # Battle of Arica & Flag Day (often observed together)
    '2026-06-29',  # Saints Peter and Paul
    '2026-07-23',  # Air Force Day
    '2026-07-28',  # Independence Day
    '2026-07-29',  # Independence Day (second day)
    '2026-08-06',  # Battle of Junín
    '2026-08-30',  # Santa Rosa de Lima
    '2026-10-08',  # Battle of Angamos
    '2026-11-01',  # All Saints’ Day
    '2026-12-08',  # Immaculate Conception
    '2026-12-09',  # Battle of Ayacucho
    '2026-12-25',  # Christmas Day
])


# Ensure the date columns are date type
df_act['BAC'] = df_act['BAC'].astype(float)
df_act['Start_Date'] = pd.to_datetime(df_act['Start_Date'] , errors='coerce')
df_act['End_Date'] = pd.to_datetime(df_act['End_Date'] , errors='coerce')
df_act['Today'] = pd.to_datetime(df_act['Today'] , errors='coerce')

# Ensure that the start date is greater or equals than the start of next month
df_act['Start_Date'] = df_act.apply(lambda row: row['Start_Date'] if row['Start_Date'] > row['Today'] else row['Today'], axis=1)

# Ensure that the end date is greater or equals than the start of next month
df_act['End_Date'] = df_act.apply(lambda row: row['End_Date'] if row['End_Date'] > row['Today'] else row['Today'], axis=1)


#df, cost_column, start_column, end_column = df_CP_EV_togo, 'Etc', 'Start_mapping','End_mapping'
df, cost_column, start_column, end_column = df_act, 'BAC', 'Start_Date','End_Date'

df[end_column] = pd.to_datetime(df[end_column], errors='coerce')
df[start_column] = pd.to_datetime(df[start_column], errors='coerce')
df['Duration'] = (df[end_column] - df[start_column]).dt.days + 1

# Build the effective working-date range first so holidays only redistribute value.
df['Date_Range'] = df.apply(
  lambda row: [
    d for d in pd.date_range(start=row[start_column], end=row[end_column], freq='D').tolist()
    if d not in public_holidays
  ],
  axis=1
)
df['Effective_Days'] = df['Date_Range'].str.len()
df = df[df['Effective_Days'] > 0].copy()

# Allocate BAC across effective days so the monthly rollup preserves the total.
df['PV per day'] = df[cost_column] / df['Effective_Days']

# Define the step as a timedelta of 1 day
step = timedelta(days=1)

# Explode the 'Date_Range' column into separate rows
df_exploded = df.explode('Date_Range')

# Add a new column 'Start_Month' with the start of the month for each date in 'Date_Range'
df_exploded['Start_Month'] = df_exploded['Date_Range'].dt.to_period('M').dt.to_timestamp()
df_exploded


# Group by 'Activity_ID' and 'Start_Month' to sum the effective daily value
df_monthly = df_exploded.groupby(['COA','Activity_ID',  'Activity_Description','Start_Month'], as_index=False)['PV per day'].sum()
df_monthly

df_monthly ['evm_type'] = 'pv'

df_monthly= df_monthly.rename(columns={
    'Start_Month': 'Data_Date',
    'PV per day': 'EVM_value'
})

df_monthly['EVM_value'].sum()

#export to excel
#df_monthly.to_excel(r'C:\Users\efloresmendoza\OneDrive - John Holland Group\2024\140_Erick\01_database\01.-COST ACCOUNTING\df_monthly.xlsx', index=False)
```

```python
df_monthly_pv = df_monthly.groupby(['COA','Data_Date', 'evm_type'], as_index=False)['EVM_value'].sum()
df_monthly_pv
```

# Compute PV cummulative by coa and period

```python
# unpivot the DataFrame to have 'evm_type' as columns
df_pivot = df_monthly_pv.pivot_table(index=['COA', 'Data_Date'], columns='evm_type', values='EVM_value', fill_value=0).reset_index()
df_pivot
```

```python
# compute pv cumulative
project_end_date = df_pivot['Data_Date'].max()
full_dates = pd.date_range(df_pivot['Data_Date'].min(), project_end_date, freq='MS')
all_coas = df_pivot['COA'].drop_duplicates()
full_index = pd.MultiIndex.from_product([all_coas, full_dates], names=['COA', 'Data_Date'])
df_pivot = (
    df_pivot.set_index(['COA', 'Data_Date'])
    .reindex(full_index, fill_value=0)
    .reset_index()
)
df_pivot['PV_Cumulative'] = df_pivot.groupby('COA')['pv'].cumsum()

# rename Data_Date to Period
df_pivot = df_pivot.rename(columns={'Data_Date': 'Period'})
df_pivot
```

```python
import sqlite3

table_name = "COA_pv_cumulative_table"
print(table_name)

# Connect
conn = sqlite3.connect(r'C:\Users\efloresmendoza\OneDrive - John Holland Group\2024\140_Erick\01_database\01.-COST ACCOUNTING\ROsample.db')
cursor = conn.cursor()

# Drop table
cursor.execute(f"DROP TABLE IF EXISTS {table_name}")

# Write DataFrame
df_pivot.to_sql(table_name, conn, if_exists='replace', index=False)

# Close
conn.close()
```
