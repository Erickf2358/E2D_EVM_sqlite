```python
df_merged = pd.merge(grouped, df_PV, left_on=['COA', 'Period'], right_on=['COA', 'Period'], how='left')
df_merged
```

```python
df_merged['CV']= df_merged['EV'] - df_merged['AC']

df_merged['SV']= df_merged['EV'] - df_merged['PV_Cumulative']
df_merged['CPI']= df_merged['EV'] / df_merged['AC']
df_merged['SPI']= df_merged['EV'] / df_merged['PV_Cumulative']

df_merged
```

```python
import pandas as pd
import sqlite3
from datetime import timedelta  # <-- needed for date_range with Timedelta or you can use freq='D'

conn = sqlite3.connect(r'C:\Users\efloresmendoza\OneDrive - John Holland Group\2024\140_Erick\01_database\01.-COST ACCOUNTING\ROsample.db')

# Define the SQL query
query_04 = """
SELECT *
FROM period_2026_05

"""

df = pd.read_sql_query(query_04, conn)
conn.close()

conn = sqlite3.connect(r'C:\Users\efloresmendoza\OneDrive - John Holland Group\2024\140_Erick\01_database\01.-COST ACCOUNTING\ROsample.db')

# Define the SQL query
query_04A = """
SELECT *
FROM period_2026_06

"""

dfA = pd.read_sql_query(query_04A, conn)
conn.close()
appended_df = pd.concat([df, dfA], ignore_index=True)
```

Work with Earned Value

```python
df_ev = appended_df[['COA', 'Activity', 'Start', 'Finish', 'EV','AC','Period']]

# Aggregate EV and AC by COA and Period
df_evA = df_ev.groupby(['COA', 'Period']).agg({'EV': 'sum', 'AC': 'sum'}).reset_index()

# Calculate the difference between the current and previous period for EV and AC
df_evA['EV_shifted'] = df_evA.groupby('COA')['EV'].shift(1).fillna(0)
df_evA['AC_shifted'] = df_evA.groupby('COA')['AC'].shift(1).fillna(0)


df_evA['EV_period'] = df_evA['EV']- df_evA['EV_shifted']
df_evA['AC_period'] = df_evA['AC']- df_evA['AC_shifted']

df_evB = df_evA[['COA', 'Period', 'EV_period', 'AC_period']]

melted = df_evB.melt(id_vars=['COA', 'Period'], value_vars=['EV_period', 'AC_period'], var_name='Metric', value_name='Value')
melted
```

```python
import pandas as pd
import sqlite3

conn = sqlite3.connect(r'C:\Users\efloresmendoza\OneDrive - John Holland Group\2024\140_Erick\01_database\01.-COST ACCOUNTING\ROsample.db')

# Define the SQL query
query_05 = """
SELECT *
FROM COA_monthly_pv_table
"""
df_pv_phased = pd.read_sql_query(query_05, conn)
conn.close()
# rename columns
df_pv_phased.rename(columns={
    'Data_Date': 'Period',
    'evm_type': 'Metric',
    'EVM_value': 'Value'
}, inplace=True)

# Append Dataframes

df_combined = pd.concat([melted, df_pv_phased], ignore_index=True)

df_combined 
```
