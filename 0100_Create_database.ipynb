```python
import pandas as pd
# File path
file_path = r'C:\Users\efloresmendoza\OneDrive - John Holland Group\2024\140_Erick\01_database\01.-COST ACCOUNTING'

# Files
file_path_CP_2 = 'RO.xlsx'

# Full import statement
sheet_mapping = 'Budget Cost'
df_project = pd.read_excel(fr'{file_path}\{file_path_CP_2}', sheet_name=sheet_mapping )

# make first row as header
df_project.columns = df_project.iloc[0]
df_project = df_project[1:]

# remove the last 6 columns
df_project = df_project.iloc[:, :-6]


# remove missing value in the column cantidad
df_project = df_project[df_project['Cantidad'].notna()]

df_project
```

```python
df = df_project.copy()  
import re
# Clean the headers
def clean_headers(headers):
    cleaned_headers = []
    for header in headers:
        # Replace non-letter characters with underscores
        cleaned_header = re.sub(r'[^a-zA-Z0-9]', '_', header)
        cleaned_headers.append(cleaned_header)
    return cleaned_headers

# Get the original headers
original_headers = df.columns.tolist()

# Clean the headers
cleaned_headers = clean_headers(original_headers)

# Rename the columns
df.columns = cleaned_headers

# Connect to SQLite database (or create it if it doesn't exist)
# Load a sheet into a DataFrame
df
```
