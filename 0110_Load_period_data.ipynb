Create Period table for EV AC and ETC

Declare period

```python
import pandas as pd

period = pd.to_datetime('01/05/2026', dayfirst=True)
# File path
file_path = r'C:\Users\efloresmendoza\OneDrive - John Holland Group\2024\140_Erick\01_database\01.-COST ACCOUNTING'

# Files
file_path_CP_2 = 'RO.xlsx'

# Full import statement
sheet_mapping = 'Activtiy Progress'
df_period = pd.read_excel(fr'{file_path}\{file_path_CP_2}', sheet_name=sheet_mapping )
df_period['Period'] = period
df_period

#Rename Cost account column to match the COA_pv_cumulative_table
df_period.rename(columns={'Cost account': 'COA'}, inplace=True)

import sqlite3

table_name = f"period_{period.year}_{period.month:02d}"
print(table_name)

# Connect
conn = sqlite3.connect(r'C:\Users\efloresmendoza\OneDrive - John Holland Group\2024\140_Erick\01_database\01.-COST ACCOUNTING\ROsample.db')
cursor = conn.cursor()

# Drop table
cursor.execute(f"DROP TABLE IF EXISTS {table_name}")

# Write DataFrame
df_period.to_sql(table_name, conn, if_exists='replace', index=False)

# Close
conn.close()
```
